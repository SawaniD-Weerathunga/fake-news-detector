```python
import pandas as pd

df_true = pd.read_csv("../data/raw/True.csv")
df_fake = pd.read_csv("../data/raw/Fake.csv")

df_true["label"] = 1
df_fake["label"] = 0

df = pd.concat([df_true, df_fake], ignore_index=True)
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

df.head()
```

```python
from src.preprocessing.text_cleaner import TextCleaner, build_tfidf_vectorizer, save_vectorizer
```

```python
df = df.dropna(subset=["text"])  # remove rows where original text is missing
```

```python
cleaner = TextCleaner()
df["clean_text"] = df["text"].apply(cleaner.clean)

vectorizer = build_tfidf_vectorizer()
X = vectorizer.fit_transform(df["clean_text"])

save_vectorizer(vectorizer)
```

```python
df["clean_text"] = df["clean_text"].fillna("")   # or df.dropna(subset=["clean_text"], inplace=True)
```
